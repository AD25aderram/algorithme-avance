```python
import random
import time

def suffixe(chp, CH):
    return CH.endswith(chp)

def CALCUL_FONCTION_TRANSITION(pattern, alphabet):
    m = len(pattern)
    transition = [{c: 0 for c in alphabet} for _ in range(m+1)]
    
    for q in range(m+1):
        for c in alphabet:
            k = min(m, q + 1)
            while k > 0 and not suffixe(pattern[:k], pattern[:q] + c):
                k -= 1
            transition[q][c] = k
    return transition

def RECHERCHE_AUTOMATE_FINI(text, pattern, alphabet):
    transition = CALCUL_FONCTION_TRANSITION(pattern, alphabet)
    state = 0
    count = 0
    m = len(pattern)
    
    for char in text:
        if char not in alphabet:
            state = 0
        else:
            state = transition[state][char]
        if state == m:
            count += 1
    return count

def RECHERCHE_NAIVE(text, pattern):
    count = 0
    m = len(pattern)
    n = len(text)
    for i in range(n - m + 1):
        if text[i:i+m] == pattern:
            count += 1
    return count



alphabet = ['a', 'b', 'c']
    

motif_length = int(input("Entrez la longueur du motif : "))
texte_length = int(input("Entrez la longueur du texte : "))
pattern = input(f"Entrez le motif sur l'alphabet {alphabet} : ")
    

text = ''.join(random.choice(alphabet) for _ in range(texte_length))
    
print("\nLes 100 premiers caractères du texte généré :")
print(text[:100])
    

start_naive = time.time()
count_naive = RECHERCHE_NAIVE(text, pattern)
end_naive = time.time()
    
print(f"\nNombre d'occurrences (méthode naïve) : {count_naive}")
print(f"Temps d'exécution (méthode naïve) : {end_naive - start_naive:.6f} secondes")
    

start_automate = time.time()
count_automate = RECHERCHE_AUTOMATE_FINI(text, pattern, alphabet)
end_automate = time.time()
    
print(f"\nNombre d'occurrences (méthode automate fini) : {count_automate}")
print(f"Temps d'exécution (méthode automate fini) : {end_automate - start_automate:.6f} secondes")
```
