```python
import time


def FiboRN(n):
    if n == 0 or n == 1:
        return n
    return FiboRN(n-1) + FiboRN(n-2)


memo = {}
def FiboRMem(n):
    if n in memo:
        return memo[n]
    if n == 0 or n == 1:
        memo[n] = n
        return n
    memo[n] = FiboRMem(n-1) + FiboRMem(n-2)
    return memo[n]


n = 35  

t1 = time.time()
print("Fibo récursif naïf =", FiboRN(n))
t2 = time.time()

t3 = time.time()
print("Fibo mémoïsation   =", FiboRMem(n))
t4 = time.time()

print("\nTemps récursif naïf :", t2 - t1, "secondes")
print("Temps mémoïsation   :", t4 - t3, "secondes")
```

```python
Monnaie = [1,2,5,10,20,50,100,200]

def PRM_recN(s):
    if s == 0:
        return 0, []  # 0 pièces – solution vide

    solutions = []

    for m in Monnaie:
        if m <= s:
            n, sol = PRM_recN(s - m)
            solutions.append((1 + n, sol + [m]))
        else:
            break

    return min(solutions, key=lambda x: x[0])


memo = {}  

def PRM_recD(s):
    if s in memo:
        return memo[s]

    if s == 0:
        memo[s] = (0, [])
        return memo[s]

    solutions = []

    for m in Monnaie:
        if m <= s:
            n, sol = PRM_recD(s - m)
            solutions.append((1 + n, sol + [m]))
        else:
            break

    memo[s] = min(solutions, key=lambda x: x[0])
    return memo[s]


def PRM_iter(s):
    T = [0] * (s+1)          # n_min
    Sol = [[] for _ in range(s+1)]  # solution optimale

    for i in range(1, s+1):
        best = (10**9, [])   # grand nombre
        for m in Monnaie:
            if m <= i:
                candidate = (1 + T[i-m], Sol[i-m] + [m])
                if candidate[0] < best[0]:
                    best = candidate
            else:
                break
        T[i] = best[0]
        Sol[i] = best[1]

    return T[s], Sol[s]



def PRM_iterD(s):
    dico = {0: (0, [])}  # s : (n_min, solution)

    for i in range(1, s+1):
        best = (10**9, [])
        for m in Monnaie:
            if m <= i:
                prev_n, prev_sol = dico[i-m]
                candidate = (1 + prev_n, prev_sol + [m])
                if candidate[0] < best[0]:
                    best = candidate
            else:
                break

        dico[i] = best

    return dico[s]


print(PRM_recD(73))
print(PRM_iter(73))
print(PRM_iterD(73))
```

```python
import time

def lev_distance_optimized(a, b):
    la, lb = len(a), len(b)

    if la > lb:
        a, b = b, a
        la, lb = lb, la

    previous = list(range(lb+1))   # ligne i-1
    current = [0] * (lb+1)         # ligne i

    for i in range(1, la+1):
        current[0] = i
        for j in range(1, lb+1):
            cost = 0 if a[i-1] == b[j-1] else 1

            current[j] = min(
                previous[j] + 1,      # suppression
                current[j-1] + 1,     # insertion
                previous[j-1] + cost  # substitution
            )

        previous, current = current, previous

    return previous[lb]

def levRec(a,b):
    if len(a) == 0:
        return len(b)
    if len(b) == 0:
        return len(a)
    if a[0] == b[0]:
        return levRec(a[1:], b[1:])
    return 1 + min(
        levRec(a[1:], b),
        levRec(a, b[1:]),
        levRec(a[1:], b[1:])
    )

# chaînes longues
s1 = "A"*12 + "B"*12
s2 = "A"*12 + "C"*12

# solution 1 
t1 = time.time()
levRec(s1, s2)  
t2 = time.time()

# solution 2 optimisée
t3 = time.time()
lev_distance_optimized(s1, s2)
t4 = time.time()

print("Temps solution récursive naïve :", t2 - t1)
print("Temps solution optimisée       :", t4 - t3)
```
